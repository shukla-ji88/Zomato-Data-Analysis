```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="listed_in(type)",
    order=df["listed_in(type)"].value_counts().index
)

plt.title("Number of Restaurants by Type")
plt.xlabel("Restaurant Type")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Clean restaurant ratings
df["rate_clean"] = (
    df["rate"]
    .astype(str)
    .str.replace("/5", "", regex=False)
    .str.strip()
)

df["rate_clean"] = pd.to_numeric(
    df["rate_clean"], errors="coerce"
)

# Compare average ratings
print(df.groupby("online_order")["rate_clean"].mean())
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x="online_order",
    y="rate_clean",
    errorbar=None
)

plt.title("Average Ratings: Online vs Offline")
plt.xlabel("Online Orders")
plt.ylabel("Average Rating")
plt.ylim(0, 5)
plt.show()
```

```python
df["approx_cost(for two people)"].value_counts().head(10)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="approx_cost(for two people)",
    bins=20
)

plt.title("Distribution of Restaurant Cost for Two")
plt.xlabel("Approximate Cost for Two")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
cost = pd.to_numeric(
    df["approx_cost(for two people)"],
    errors="coerce"
)

print("Average cost for two:", cost.mean())
print("Median cost for two:", cost.median())
```

```python
df.groupby("listed_in(type)")["votes"].sum().sort_values(ascending=False)
```

```python
votes_by_type = (
    df.groupby("listed_in(type)")["votes"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=votes_by_type.index,
    y=votes_by_type.values
)

plt.title("Total Votes by Restaurant Type")
plt.xlabel("Restaurant Type")
plt.ylabel("Total Votes")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
order_table = pd.crosstab(
    df["listed_in(type)"],
    df["online_order"]
)

print(order_table)
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(
    order_table,
    annot=True,
    fmt="d",
    cmap="YlGnBu"
)

plt.title("Restaurant Type vs Online Orders")
plt.xlabel("Online Order")
plt.ylabel("Restaurant Type")
plt.tight_layout()
plt.show()
```

```python
# Zomato Data Analysis - Final Summary

print("===== ZOMATO DATA ANALYSIS =====")

print("\n1. Total restaurants:", len(df))

print("\n2. Restaurant types:")
print(df["listed_in(type)"].value_counts())

print("\n3. Online order distribution:")
print(df["online_order"].value_counts())

print("\n4. Average rating by online order:")
print(df.groupby("online_order")["rate_clean"].mean().round(2))

print("\n5. Average cost for two:")
print(round(cost.mean(), 2))

print("\n6. Total votes by restaurant type:")
print(
    df.groupby("listed_in(type)")["votes"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
summary = pd.DataFrame({
    "Metric": [
        "Total Restaurants",
        "Average Cost for Two",
        "Highest Restaurant Type by Count",
        "Highest Restaurant Type by Votes"
    ],
    "Result": [
        len(df),
        round(cost.mean(), 2),
        df["listed_in(type)"].value_counts().idxmax(),
        df.groupby("listed_in(type)")["votes"].sum().idxmax()
    ]
})

summary.to_csv("Zomato_Analysis_Report.csv", index=False)

print("Report saved successfully!")
```
